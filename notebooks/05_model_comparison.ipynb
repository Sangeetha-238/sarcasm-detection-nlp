```python
import pandas as pd

log_reg = pd.read_csv(
    "../results/logistic_regression_metrics.csv"
)

svm = pd.read_csv(
    "../results/svm_metrics.csv"
)

comparison = pd.concat(
    [log_reg, svm],
    ignore_index=True
)

comparison
```

# Classical Model Comparison

This notebook compares the performance of the two classical machine learning models used for sarcasm detection:

* Logistic Regression
* Linear Support Vector Machine (SVM)

Both models were trained using the same TF-IDF representation and the same train/validation split. This ensures that the comparison is fair and that any difference in performance is due to the classifier rather than differences in preprocessing or data.

---

## Evaluation Metrics

The models are compared using the following evaluation metrics:

* **Accuracy** — overall proportion of correct predictions
* **Precision** — proportion of predicted sarcastic headlines that were actually sarcastic
* **Recall** — proportion of sarcastic headlines correctly identified
* **F1-score** — harmonic mean of precision and recall

F1-score is especially useful because it provides a balanced measure of classification performance.

---

## Model Performance

Both Logistic Regression and Linear SVM achieved approximately **85% validation accuracy**.

### Logistic Regression

* True Negatives: **1,889**
* False Positives: **353**
* False Negatives: **303**
* True Positives: **1,730**

### Linear SVM

* True Negatives: **1,894**
* False Positives: **348**
* False Negatives: **300**
* True Positives: **1,733**

Although the classification reports appear almost identical when rounded to two decimal places, Linear SVM made slightly fewer classification errors overall.

---

## Confusion Matrix Comparison

The Logistic Regression model produced:

* **656 total classification errors**
* 353 false positives
* 303 false negatives

The Linear SVM model produced:

* **648 total classification errors**
* 348 false positives
* 300 false negatives

Linear SVM therefore made **8 fewer errors** on the validation set.

The difference is small, which indicates that both classifiers perform similarly when using the same TF-IDF feature representation.

---

## Performance Interpretation

The similarity between Logistic Regression and Linear SVM suggests that the TF-IDF representation is responsible for much of the predictive performance.

Both models are linear classifiers operating on the same high-dimensional sparse feature space, so similar results are expected.

Linear SVM performs slightly better based on the confusion matrix, but the improvement is not large enough to suggest a major advantage over Logistic Regression.

This indicates that further improvements are more likely to come from using richer text representations rather than simply switching between similar linear classifiers.

---

## Classical Model Selection

Based on the validation results, **Linear SVM is selected as the strongest classical machine learning baseline**.

The decision is based on:

* Slightly fewer false positives
* Slightly fewer false negatives
* Slightly fewer total classification errors
* Comparable precision, recall, and F1-score across both classes

Because the difference between the two models is small, Logistic Regression remains a strong baseline and provides useful interpretability through its feature coefficients.

---

## Key Takeaway

Traditional machine learning approaches using TF-IDF achieve strong performance on this sarcasm detection dataset.

However, TF-IDF primarily captures lexical patterns and short word combinations. It does not explicitly model word context, semantic meaning, or relationships between words.

Sarcasm often depends on contextual or semantic contradictions that may not be captured by bag-of-words representations.

The next stage of the project will therefore explore a **transformer-based model** and evaluate whether contextual language representations improve sarcasm detection performance.
